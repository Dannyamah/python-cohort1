### Error Handling

#### Exception (Try/Except)

try: Run the code

except: If it crashes, ignore error silently or do something else

finally: Clean up no matter what happens

```python
## Error Handling Using Try/Except 

try:
    with open('none.file.txt', 'r') as file:
        content = file.read()

except:
    pass

finally:
    print('This will always execute regardless of error')
```

### Common Errors


```python
# ValueError

number = 'ten'
value = int(number)
print("Caught a TypeError: Cannot add string and integer")
```

```python
## ValueError Handling

number = 'ten'

try:
    value = int(number)
except ValueError:
    print("Caught a TypeError: Cannot add string and integer")
```

```python
# TypeError 

result = '5' + 5

print("Caught a TypeError: Cannot add string and integer")
```

```python
# TypeError Handling

try:
    result = '5' + 5
except TypeError:
    print("Caught a TypeError: Cannot add string and integer")
```

```python
try:
    with open('wrongfile.txt', 'r') as file:
        content = file.read()
except:
    print("Caught a FileNotFoundError: The File does not exist")
```

```python
# KeyError Handling

my_dict = {'a':1, 'b':2}

try:
    value = my_dict['c']
except KeyError:
    print("Caught a KeyError: The key does not exist in the dictionary")
```

```python
# IndexError Handling

my_list = [1,2,3]

try:
    value = my_list[5]
except IndexError:
    print("Caught an IndexError: The index is out of range")
```
